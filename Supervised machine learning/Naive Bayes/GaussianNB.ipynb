# Breast Cancer Classification with Gaussian Naive Bayes
In this notebook, we build a simple **Gaussian Naive Bayes** classifier to predict whether a breast tumor is malignant or benign, using the built-in `load_breast_cancer` dataset from scikit-learn.

## Libraries

```python
import matplotlib.pyplot as plt
import  seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix ,classification_report , accuracy_score
```

## Load Dataset

```python
df=load_breast_cancer()
```

```python
X=df.data
```

```python
X.shape
```

```python
y=df.target
```

## Train/Test Split

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y, train_size=0.8,random_state=42
)
```

## Model Training

```python
GaussianNBModel=GaussianNB()
```

```python
GaussianNBModel.fit(X_train,y_train)
```

```python
y_train_pred = GaussianNBModel.predict(X_train)
y_test_pred = GaussianNBModel.predict(X_test)
y_pred_prob = GaussianNBModel.predict_proba(X_test)
```

## Evaluate

```python
train_accuracy = accuracy_score(y_train, y_train_pred)
test_accuracy = accuracy_score(y_test, y_test_pred)
print(f'{train_accuracy:.2f} , {test_accuracy:.2f}')
```

```python
print(classification_report(y_train,y_train_pred), '\n',classification_report(y_test,y_test_pred))
```

```python
train_accuracy = GaussianNBModel.score(X_train, y_train)
test_accuracy = GaussianNBModel.score(X_test, y_test)

train_accuracy ,test_accuracy
```

## Confusion Matrix

```python
CM=confusion_matrix(y_test,y_test_pred)
sns.heatmap(
    CM,
    annot=True,
    fmt='g',
    cmap='Blues',
)
plt.xlabel('Prediction')
plt.ylabel('Actual')
```

## Conclusion

The Gaussian Naive Bayes model achieves **94% accuracy on the training set** and **97% accuracy on the test set**, with balanced precision and recall across both classes. This shows the model generalizes well on this dataset.
